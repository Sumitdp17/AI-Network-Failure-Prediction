```python
from src.data.load_data import load_data
from src.features.preprocessing import DataPreprocessor
```

```python
train_df, test_df = load_data(
    "../data/raw/UNSW_NB15_training-set.csv",
    "../data/raw/UNSW_NB15_testing-set.csv"
)
```

```python
preprocessor = DataPreprocessor()

train_df = preprocessor.drop_unnecessary_columns(train_df)
test_df = preprocessor.drop_unnecessary_columns(test_df)
```

```python
X_train, y_train = preprocessor.split_features_target(train_df)
X_test, y_test = preprocessor.split_features_target(test_df)
```

```python
X_train, X_test = preprocessor.encode_features(
    X_train,
    X_test
)
```
